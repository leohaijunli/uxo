# 3-D GPR Migration Inversion Report

**Multi-modal (GPR + magnetometer) UXO dataset — OSU seeded set.**

This notebook inverts the simulated **3-D GPR survey** for each object: a 2-D antenna grid of B-scans (dx=dy=0.02 m from the real drone speed/PRF, 0.05 m ground-coupled height, gprMax 3-D, gprMax LandmineModels where applicable) is background-subtracted and **Kirchhoff-migrated** into a 3-D image I(x,y,z). The target is located at the image peak -> full 3-D position (x, y, depth).

**GPR vs magnetometry (why both):** GPR sees *electrical* contrast (any metal = PEC, plus plastic/void/TNT dielectric contrast); magnetometry sees *magnetic* content. They are orthogonal: e.g. D22 (plastic mine + steel mechanism) is hard for GPR but easy for magnetometry; F7 (aluminum can) is easy for GPR but invisible to magnetometry.

**Drag to rotate** the 3-D volume plots below (plotly, loaded from CDN).
